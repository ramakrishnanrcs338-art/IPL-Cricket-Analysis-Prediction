## Basic Cleaning

We should **not** blindly replace every missing value with zero. For example, a missing `player_of_match` is different from a numeric run value of zero.

For the initial cleaning:
- Convert `date` to datetime.
- Keep missing player/umpire fields as missing unless they are required for a particular analysis.
- Fill missing numeric delivery fields with 0 where 0 is the meaningful absence of that event.
- Remove exact duplicate rows.


```python
# Make copies before cleaning
matches_clean = matches.copy()
deliveries_clean = deliveries.copy()

# Convert date
if "date" in matches_clean.columns:
    matches_clean["date"] = pd.to_datetime(matches_clean["date"], errors="coerce")

# Remove exact duplicates
matches_clean = matches_clean.drop_duplicates().reset_index(drop=True)
deliveries_clean = deliveries_clean.drop_duplicates().reset_index(drop=True)

# Numeric delivery columns where missing means no runs/event
delivery_numeric_cols = [
    "wide_runs", "bye_runs", "legbye_runs", "noball_runs",
    "penalty_runs", "batsman_runs", "extra_runs", "total_runs"
]

for col in delivery_numeric_cols:
    if col in deliveries_clean.columns:
        deliveries_clean[col] = deliveries_clean[col].fillna(0)

print("Clean matches shape:", matches_clean.shape)
print("Clean deliveries shape:", deliveries_clean.shape)
```

```python
# Quick sanity checks after cleaning
print("Date range:", matches_clean["date"].min(), "to", matches_clean["date"].max())
print("Seasons:", sorted(matches_clean["season"].dropna().unique()))
print("Number of teams:", len(set(matches_clean["team1"].dropna()) | set(matches_clean["team2"].dropna())))
print("Number of unique venues:", matches_clean["venue"].nunique())
```

```python
# Team names used in the dataset
teams = sorted(set(matches_clean["team1"].dropna()) | set(matches_clean["team2"].dropna()))
print("Teams:")
for team in teams:
    print("-", team)
```
